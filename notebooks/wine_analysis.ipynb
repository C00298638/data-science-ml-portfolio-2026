```python
import matplotlib.pyplot as plt

df["quality"].value_counts().sort_index().plot(
    kind="bar",
    xlabel="Wine Quality Score",
    ylabel="Number of Wines",
    title="Distribution of Red Wine Quality Scores"
)

plt.tight_layout()
plt.show()
```

The different wine quality scores and which ones appear most often and least often.

```python
df["alcohol"].hist(bins=20)

plt.xlabel("Alcohol (% by volume)")
plt.ylabel("Number of Wines")
plt.title("Distribution of Alcohol Content")

plt.tight_layout()
plt.show()
```
